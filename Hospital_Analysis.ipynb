```python
import pandas as pd
import numpy as np
```

```python
patients = pd.read_csv("patients.csv")
encounters = pd.read_csv("encounters.csv")
procedures = pd.read_csv("procedures.csv")
payers = pd.read_csv("payers.csv")
organizations = pd.read_csv("organizations.csv")
```

```python
datasets = {
    "patients": patients,
    "encounters": encounters,
    "procedures": procedures,
    "payers": payers,
    "organizations": organizations
}

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("-"*50)
    print("Shape:", df.shape)
    print("Missing Values:")
    print(df.isnull().sum())
    print("\nDuplicates:", df.duplicated().sum())
```

```python
patients['MARITAL'] = patients['MARITAL'].fillna('Unknown')
```

```python
patients['BIRTHDATE'] = pd.to_datetime(patients['BIRTHDATE'])
patients['DEATHDATE'] = pd.to_datetime(patients['DEATHDATE'])
```

```python
from datetime import datetime

today = pd.Timestamp.today()

patients['AGE'] = (
    (today - patients['BIRTHDATE']).dt.days // 365
)
```

```python
patients['ZIP'] = patients['ZIP'].fillna('Unknown')
```

```python
encounters['REASONCODE'] = encounters['REASONCODE'].fillna('Unknown')
encounters['REASONDESCRIPTION'] = encounters['REASONDESCRIPTION'].fillna('Unknown')
```

```python
encounters['START'] = pd.to_datetime(encounters['START'])
encounters['STOP'] = pd.to_datetime(encounters['STOP'])
```

```python
encounters['LENGTH_OF_STAY_DAYS'] = (
    encounters['STOP'] - encounters['START']
).dt.total_seconds() / (24 * 60 * 60)
```

```python
encounters[['START','STOP','LENGTH_OF_STAY_DAYS']].head()
```

```python
print("Total Encounters:", len(encounters))

print("Average Stay:",
      round(encounters['LENGTH_OF_STAY_DAYS'].mean(),2),
      "days")

print("Average Claim Cost:",
      round(encounters['TOTAL_CLAIM_COST'].mean(),2))

print("Average Coverage:",
      round(encounters['PAYER_COVERAGE'].mean(),2))
```
